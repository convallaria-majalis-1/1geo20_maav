PREGUNTA 2: Renombrar la columna "CÓDIGO" a "mes" y elimina la columna “PROM. TOTAL”

```python
df=df.rename(columns= {'CÓDIGO':'mes'})
df=df.drop(columns= {'PROM. TOTAL'})
df
```

```python
df.head()
```

```python
# 1. Transformo la tabla de formato ancho a largo
df_largo = df.melt(
    id_vars=['mes', 'Día'],
    var_name='anio',
    value_name='valor'
)

# 2. Defino el tipo de objeto
df_largo['anio'] = df_largo['anio'].astype(str)
df_largo['mes'] = df_largo['mes'].astype(str).str.zfill(2)
df_largo['Día'] = df_largo['Día'].astype(str).str.zfill(2)
```

```python
# 3. Creo la columna de fecha (29/02 de años no bisiestos se convertirán en NaT)
df_largo['fecha'] = pd.to_datetime(
    df_largo['Día'] + '-' + df_largo['mes'] + '-' + df_largo['anio'],
    format='%d-%m-%Y',
    errors='coerce'
)

# 4. Elimino fechas no válidas
df_resultado = (
    df_largo
    .dropna(subset=['fecha'])
    [['fecha', 'valor']]
    .sort_values('fecha')
    .reset_index(drop=True)
)

print(df_resultado)
```

Para evaluar que los valores inválidos (año no bisiesto) para 29-02 se han eliminado correctamente:

```python
feb_29_entry = df_resultado[
    (df_resultado['fecha'].dt.month == 2) &
    (df_resultado['fecha'].dt.day == 29)
]

print("Valores para 29-02:\n\n", feb_29_entry)
```
